## 🧱 核心知識點一：官方驅動庫的演進 (標準庫 vs HAL 庫)
當我們想控制 STM32 的硬體（例如讓一個引腳輸出高電位）時，我們有兩種官方提供的工具箱：

1. **標準外設庫 (Standard Peripheral Library, 簡稱標準庫/SPL)**：
   * **本質**：直接對暫存器（Register）進行簡單函數封裝。
   * **命運**：官方已停止維護。其最大致命傷是**移植性極差**（STM32F1 的程式碼無法直接用在 STM32F4）。
2. **HAL 庫 (Hardware Abstraction Layer, 硬體抽象層)**：
   * **本質**：現代 STM32 生態系（STM32Cube）的核心。它將硬體徹底抽象化，管你底層是哪顆晶片，控制引腳的函數都統一叫 `HAL_GPIO_Init()`。
   * **優勢**：極佳的跨晶片移植性，且完美支援圖形化配置工具（STM32CubeMX）。

> **📌 學生思維升級（BSP 概念）**
> 在實際專案中，我們會在 HAL 庫之上再封裝一層 **BSP (Board Support Package，板級支持包)**。
> * **HAL 庫** 關心晶片（「把 GPIOA Pin 5 拉高」）。
> * **BSP 庫** 關心板子（「晶片 GPIOA Pin 5 連接的是綠色 LED」）。
> 透過這種分層，你的主程式（App）只需要呼叫 `BSP_LED_Green_Toggle()`。未來更換開發板，主程式一個字都不用改，這就是軟體工程的優雅美學。

## 🎬 核心知識點二：晶片通電後的開機真相 (裸機啟動流程)
CPU 醒來的第一秒，根本不知道什麼是 C 語言的 `main()` 函數。它是如何一步步走到你寫的程式碼呢？這全靠**啟動文件 (Startup File, `.s` 檔)** 在幕後導引。



### 裸機啟動的五大晶片原力：
1. **中斷向量表 (Vector Table)**：晶片通電時，硬體會自動去快閃記憶體（Flash）的開頭讀取清單。這張清單就寫在啟動文件（`.s`）的開頭，記載了「堆疊指針（MSP）」和「復位向量（Reset）」。
2. **進入 Reset_Handler**：拿到復位向量後，CPU 開始執行啟動文件裡第一段組合語言程式碼。
3. **調用 SystemInit**：啟動文件會去呼叫 C 語言寫的系統時鐘初始化函數，把晶片頻率調校到正常狀態（例如 72MHz）。
4. **記憶體搬運工**：在進入 `main` 之前，啟動文件必須把躺在 Flash 裡有初始值的全域變數複製到執行記憶體（SRAM）中（此區域稱為 `.data` 段），並把未初始化的全域變數區域全部刷成 0（稱為 `.bss` 段）。
5. **奔向 main()**：萬事俱備，執行組合語言指令 `bl main`，世界正式交給你的 C 語言程式。

> **⚠️ 觀念澄清：啟動文件（.s）vs 引導程式（Bootloader）**
> * **啟動文件**：大樓的「地基與水電」。是 C 語言運行的物理基礎，**晶片開機必吃的第一道菜**。
> * **Bootloader**：大樓的「物業管理員/電梯小姐」。負責檢查有沒有要更新韌體（如透過網路或隨身碟），若沒有更新需求，它會引導跳轉（Jump）到應用程式的啟動文件。

## 🛠️ 核心知識點三：編譯器與構建系統的黑科技 (GCC, Clang, CMake)
每一種不同型號的晶片，都有自己專屬的**記憶體地圖（`.ld` 連結器指令檔）**與**啟動檔（`.s`）**。在 Linux 系統下，是誰在管理這一切？

[你寫的C程式碼] ---> [編譯器 (GCC/Clang)] ---> [連結器 (.ld地圖)] ---> [.hex/.bin 韌體]

^                                             ^
|_______________________(由 CMake 指揮)_______|


### 1. 跨平台編譯器的革命：GCC 與 Clang
在樹莓派 Linux 上，我們可以使用傳統的 `arm-none-eabi-gcc` 工具鏈進行編譯。
但更令人驚奇的是，現代的 **Clang/LLVM** 編譯器是一個「全能編譯器」。你不需要下載專屬的嵌入式版本，只要在樹莓派原生的 `clang` 指令後面加上參數：
`clang -target arm-none-eabi -mcpu=cortex-m3 -mthumb`
Clang 就會瞬間啟動嵌入式發射器，把你寫的 C 語言翻譯成 STM32 的機器碼！

### 2. 樂團指揮官：CMake
既然編譯時要加入「點燈.c」、「HAL庫.c」、「啟動檔.s」，還要強制綁定「`.ld` 記憶體地圖檔」，指令會變得無比冗長。
**CMake** 的出現拯救了所有人。我們只需要在 `CMakeLists.txt` 文字檔中寫下幾行清晰的劇本：
* `add_executable(main "startup_stm32f103xb.s" "main.c")`（加入啟動檔）
* `target_link_options(main PRIVATE "-T.../FLASH.ld")`（綁定地圖檔）

在終端機輸入 `cmake ..` 與 `make`，CMake 就會在幕後自動把組合語言、硬體地圖、HAL 庫和 C 語言核心天衣無縫地縫合在一起。

```bash
sudo apt update
sudo apt install cmake gcc-arm-none-eabi libnewlib-arm-none-eabi vscode
步驟二：與 AI 協作生成「大樓地基」
你不需要自己手寫複雜的 CMakeLists.txt 或配置檔。直接打開 DeepSeek 或 Claude，用自然語言輸入以下 Prompt：

💡 AI 提示詞 (Prompt) 範例：
「我正在使用 Raspberry Pi (Linux ARM64) 開發 STM32F103C8T6。我想使用 HAL 庫與 arm-none-eabi-gcc 工具鏈。請幫我寫一個標準、優雅的 CMakeLists.txt 範本，必須包含 .ld 連結檔與 startup_stm32f103xb.s 啟動檔的配置，並幫我規劃好 Core、Hardware(BSP)、Drivers 的資料夾結構。」

步驟三：編譯與燒錄
AI 會為你生成完美的工程模板架構。你只需要將官方的 .s 檔和 .ld 檔放入對應資料夾，在 VS Code 中開啟，並在終端機輸入：

Bash
mkdir build && cd build
cmake ..
make -j4
幾秒鐘內，你的樹莓派就會編譯出最終的 .bin 韌體。接下來，搭配 OpenOCD 或是樹莓派的 GPIO，就能直接將程式送進 STM32 晶片中！

🎓 課堂思考題
為什麼同一顆晶片（例如 STM32F103C8），在標準庫和 HAL 庫底下的啟動文件（.s）不能直接互換混用？

試著在你的樹莓派上輸入 clang --version，查看 Registered Targets: 清單中是否包含 arm 與 thumb？這對嵌入式開發帶來什麼好處？

# 實作：點燈 main.c — 從震盪器設定到 GPIO 輸出 (HAL 庫)
**目標晶片：STM32F103C8T6 (Blue Pill) | 外接晶振：8 MHz | HCLK = 72 MHz**

本節將 step-by-step 建構一個完整的點燈程式，**從選定晶振頻率開始**，依序通過時鐘樹、Flash 延遲、外設時鐘使能、GPIO 初始化，最終讓 LED 閃爍。每一步都會說明**外設原理**與**為何這樣設定**。

> HAL 庫來源：[STMicroelectronics/STM32CubeF1](https://github.com/STMicroelectronics/STM32CubeF1) (已下載至 `/home/pi/stm32/STM32CubeF1`)

---

## Step 1 — 選擇系統時鐘源：為什麼要從 HSE 出發？

STM32F103 有三種時鐘來源可驅動系統時鐘 (SYSCLK)：

| 時鐘源 | 頻率 | 精準度 | 典型用途 |
|--------|------|--------|----------|
| **HSI** (內部高速 RC 振盪器) | 8 MHz (經工廠校准) | ±1% | 開機預設，低成本場合 |
| **HSE** (外部晶振/時鐘) | 4–16 MHz (依電路板) | ±50 ppm (晶振級) | 精準通訊 (UART/USB) |
| **PLL** (鎖相迴路倍頻) | 可程式化 (最高 72 MHz) | 取決於輸入源 | 高效能運算 |

**為何選 HSE + PLL？**
- Blue Pill 板載一顆 **8 MHz 晶振**連接到 STM32 的 OSC_IN/OSC_OUT 引腳。
- 若只用 HSI (8 MHz)，系統最高只能跑 8 MHz，無法發揮 STM32F103 的 72 MHz 效能上限；且 HSI 精度較差，用於 UART 通訊可能產生誤碼。
- 因此標準做法是：**啟用 HSE → 餵入 PLL 倍頻 → 輸出 72 MHz 作為 SYSCLK**。

> 公式：`PLL 輸出 = HSE 頻率 / HSE Prediv × PLL Mul`
> 常見配置：`8 MHz / 1 × 9 = 72 MHz`

---

## Step 2 — 啟動 HSE 並設定 PLL

HAL 庫使用 `RCC_OscInitTypeDef` 結構體來配置震盪器。

```c
/* 檔案：main.c — 完整點燈程式 (STM32F103C8, HAL 庫) */

#include "stm32f1xx_hal.h"

/* 用於 PC13 LED (Blue Pill 內建 LED 在 PC13，低電位點亮) */
#define LED_PIN         GPIO_PIN_13
#define LED_PORT        GPIOC

static void SystemClock_Config(void);
static void GPIO_Init(void);

int main(void)
{
    /* 1. HAL 庫初始化：設定 Systick 中斷優先權等底層 */
    HAL_Init();

    /* 2. 設定系統時鐘：HSE → PLL → 72 MHz */
    SystemClock_Config();

    /* 3. 初始化 LED 引腳 */
    GPIO_Init();

    /* 4. 主循環：每隔 500 ms 翻轉 LED */
    while (1)
    {
        HAL_GPIO_TogglePin(LED_PORT, LED_PIN);
        HAL_Delay(500);
    }
}

static void SystemClock_Config(void)
{
    RCC_OscInitTypeDef RCC_OscInitStruct = {0};
    RCC_ClkInitTypeDef RCC_ClkInitStruct = {0};

    /* ── 震盪器配置 ── */
    RCC_OscInitStruct.OscillatorType = RCC_OSCILLATORTYPE_HSE; /* 只操作 HSE */
    RCC_OscInitStruct.HSEState       = RCC_HSE_ON;             /* 啟動外部晶振 */
    RCC_OscInitStruct.HSEPredivValue = RCC_HSE_PREDIV_DIV1;    /* 8 MHz ÷ 1 = 8 MHz */
    RCC_OscInitStruct.PLL.PLLState   = RCC_PLL_ON;             /* 啟用 PLL */
    RCC_OscInitStruct.PLL.PLLSource  = RCC_PLLSOURCE_HSE;      /* PLL 輸入 = HSE */
    RCC_OscInitStruct.PLL.PLLMUL     = RCC_PLL_MUL9;           /* 8 MHz × 9 = 72 MHz */

    if (HAL_RCC_OscConfig(&RCC_OscInitStruct) != HAL_OK)
    {
        Error_Handler();
    }

    /* ── 系統時鐘與匯流排分頻配置 ── */
    RCC_ClkInitStruct.ClockType      = RCC_CLOCKTYPE_SYSCLK |  /* 設定 SYSCLK */
                                       RCC_CLOCKTYPE_HCLK   |  /* 設定 AHB   */
                                       RCC_CLOCKTYPE_PCLK1  |  /* 設定 APB1  */
                                       RCC_CLOCKTYPE_PCLK2;    /* 設定 APB2  */
    RCC_ClkInitStruct.SYSCLKSource   = RCC_SYSCLKSOURCE_PLLCLK;/* SYSCLK = PLL 輸出 (72 MHz) */
    RCC_ClkInitStruct.AHBCLKDivider  = RCC_SYSCLK_DIV1;        /* HCLK = SYSCLK / 1 = 72 MHz */
    RCC_ClkInitStruct.APB1CLKDivider = RCC_HCLK_DIV2;          /* PCLK1 = HCLK / 2 = 36 MHz (上限 36 MHz) */
    RCC_ClkInitStruct.APB2CLKDivider = RCC_HCLK_DIV1;          /* PCLK2 = HCLK / 1 = 72 MHz */

    /* Flash 等待週期：72 MHz 需 2 個等待週期 */
    if (HAL_RCC_ClockConfig(&RCC_ClkInitStruct, FLASH_LATENCY_2) != HAL_OK)
    {
        Error_Handler();
    }
}
```

### 為什麼 APB1 要除 2？

這是 STM32F103 的硬體限制：

| 匯流排 | 最大頻率 | 掛載外設範例 |
|--------|---------|-------------|
| **AHB** (HCLK) | 72 MHz | Flash、DMA、CPU |
| **APB2** (PCLK2) | 72 MHz | GPIO、ADC、TIM1、SPI1、USART1 |
| **APB1** (PCLK1) | **36 MHz** | TIM2~4、SPI2、I2C、USART2、RTC |

APB1 匯流排的物理設計只能跑到 36 MHz。若設為 72 MHz，匯流排通訊會不穩定。因此當 HCLK = 72 MHz 時，必須設 `RCC_HCLK_DIV2` 讓 PCLK1 = 36 MHz。

### 為什麼需要 Flash 等待週期？

Flash 記憶體讀取速度遠低於 CPU 核心。當 SYSCLK ≥ 24 MHz 時，CPU 一次指令週期內 Flash 來不及完成讀取，必須插入等待週期：

| SYSCLK 頻率 | 所需等待週期 |
|-------------|------------|
| 0 – 24 MHz | 0 (FLASH_LATENCY_0) |
| 24 – 48 MHz | 1 (FLASH_LATENCY_1) |
| 48 – 72 MHz | **2 (FLASH_LATENCY_2)** |

若等待週期不足，CPU 會讀到錯誤的指令，造成 **HardFault 或隨機崩潰**。

## Step 3 — 使能 GPIO 時鐘並設定引腳模式

在 STM32 中，**所有 GPIO 連接埠的時鐘預設是關閉的**。這是為了省電——晶片不會浪費電力去驅動沒用到的引腳電路。

```c
static void GPIO_Init(void)
{
    GPIO_InitTypeDef GPIO_InitStruct = {0};

    /* 使能 GPIOC 連接埠的時鐘 */
    __HAL_RCC_GPIOC_CLK_ENABLE();

    /* 設定 PC13 為推輓輸出，無上下拉，速度 2 MHz */
    GPIO_InitStruct.Pin   = LED_PIN;
    GPIO_InitStruct.Mode  = GPIO_MODE_OUTPUT_PP;
    GPIO_InitStruct.Pull  = GPIO_PULLUP;   /* 因 LED 低電位點亮，初始拉高避免誤閃 */
    GPIO_InitStruct.Speed = GPIO_SPEED_FREQ_LOW;  /* 2 MHz 對 LED 已足夠 */
    HAL_GPIO_Init(LED_PORT, &GPIO_InitStruct);

    /* 初始狀態：PC13 = HIGH → LED 熄滅 */
    HAL_GPIO_WritePin(LED_PORT, LED_PIN, GPIO_PIN_SET);
}

static void Error_Handler(void)
{
    /* 錯誤發生時停機，可在此設斷點 */
    while (1)
    {
        /* 可加入快速閃爍作錯誤指示 */
    }
}
```

### 引腳模式選擇分析

STM32 GPIO 每支引腳有多種模式，透過 `GPIO_InitTypeDef.Mode` 選擇：

| 模式 | 使用場景 |
|------|---------|
| `GPIO_MODE_OUTPUT_PP` (推輓輸出) | 直接驅動 LED、普通數位訊號 |
| `GPIO_MODE_OUTPUT_OD` (開漏輸出) | I²C 匯流排、位準轉換 |
| `GPIO_MODE_INPUT` (浮空輸入) | 讀取按鍵、感測器 |
| `GPIO_MODE_AF_PP` (復用推輓) | USART TX、SPI SCK 等由外設控制 |
| `GPIO_MODE_ANALOG` (類比模式) | ADC 輸入 |

**LED 點燈 → `GPIO_MODE_OUTPUT_PP`**，因為我們需要引腳輸出的電壓直接驅動 LED（透過限流電阻）。

### 速度設定為什麼選 LOW？

`GPIO_SPEED_FREQ_LOW` 對應約 2 MHz 的輸出 slew rate。LED 翻轉頻率僅 1–2 Hz，完全不需要高速模式。選用較低速度可**減少電磁干擾 (EMI)** 並**降低功耗**。

## Step 4 — 完整的 `main.c` 彙整

將上述三個步驟組合在一起：

```c
#include "stm32f1xx_hal.h"

#define LED_PIN         GPIO_PIN_13
#define LED_PORT        GPIOC

static void SystemClock_Config(void);
static void GPIO_Init(void);

int main(void)
{
    HAL_Init();
    SystemClock_Config();
    GPIO_Init();

    while (1)
    {
        HAL_GPIO_TogglePin(LED_PORT, LED_PIN);
        HAL_Delay(500);
    }
}

static void SystemClock_Config(void)
{
    RCC_OscInitTypeDef RCC_OscInitStruct = {0};
    RCC_ClkInitTypeDef RCC_ClkInitStruct = {0};

    RCC_OscInitStruct.OscillatorType = RCC_OSCILLATORTYPE_HSE;
    RCC_OscInitStruct.HSEState       = RCC_HSE_ON;
    RCC_OscInitStruct.HSEPredivValue = RCC_HSE_PREDIV_DIV1;
    RCC_OscInitStruct.PLL.PLLState   = RCC_PLL_ON;
    RCC_OscInitStruct.PLL.PLLSource  = RCC_PLLSOURCE_HSE;
    RCC_OscInitStruct.PLL.PLLMUL     = RCC_PLL_MUL9;
    if (HAL_RCC_OscConfig(&RCC_OscInitStruct) != HAL_OK)
        Error_Handler();

    RCC_ClkInitStruct.ClockType      = RCC_CLOCKTYPE_SYSCLK |
                                       RCC_CLOCKTYPE_HCLK   |
                                       RCC_CLOCKTYPE_PCLK1  |
                                       RCC_CLOCKTYPE_PCLK2;
    RCC_ClkInitStruct.SYSCLKSource   = RCC_SYSCLKSOURCE_PLLCLK;
    RCC_ClkInitStruct.AHBCLKDivider  = RCC_SYSCLK_DIV1;
    RCC_ClkInitStruct.APB1CLKDivider = RCC_HCLK_DIV2;
    RCC_ClkInitStruct.APB2CLKDivider = RCC_HCLK_DIV1;
    if (HAL_RCC_ClockConfig(&RCC_ClkInitStruct, FLASH_LATENCY_2) != HAL_OK)
        Error_Handler();
}

static void GPIO_Init(void)
{
    GPIO_InitTypeDef GPIO_InitStruct = {0};

    __HAL_RCC_GPIOC_CLK_ENABLE();

    GPIO_InitStruct.Pin   = LED_PIN;
    GPIO_InitStruct.Mode  = GPIO_MODE_OUTPUT_PP;
    GPIO_InitStruct.Pull  = GPIO_PULLUP;
    GPIO_InitStruct.Speed = GPIO_SPEED_FREQ_LOW;
    HAL_GPIO_Init(LED_PORT, &GPIO_InitStruct);

    HAL_GPIO_WritePin(LED_PORT, LED_PIN, GPIO_PIN_SET);
}

static void Error_Handler(void)
{
    while (1) { }
}
```

## 總結：時鐘樹視覺流程圖

```
┌───────────┐   ÷1      ┌──────┐   ×9      ┌───────┐   ÷1    ┌────────┐
│ HSE 8 MHz │ ───────→  │Prediv│ ───────→  │ PLL   │ ─────→ │SYSCLK  │
└───────────┘            └──────┘           │ 72 MHz│         │ 72 MHz │
                                            └───────┘         └───┬────┘
                                                                  │
                                             ┌────────────────────┤
                                             │         │          │
                                             ▼         ▼          ▼
                                         ┌──────┐ ┌──────┐  ┌────────┐
                                         │ AHB  │ │ APB2 │  │ APB1   │
                                         │ 72M  │ │ 72M  │  │ 36M    │
                                         └──────┘ └──────┘  └────────┘
```

### Key Takeaways
1. **先決定 HSE 頻率**（看電路板上晶振），再算 PLL 倍率達到目標 SYSCLK
2. **APB1 上限 36 MHz**是一個常見的新手陷阱——忘了除 2 會造成 APB1 外設異常
3. **Flash 等待週期不足**是 HardFault 的常見元兇
4. **GPIO 時鐘預設關閉**，使用前必須手動 `__HAL_RCC_GPIOC_CLK_ENABLE()`
5. **`HAL_Delay()`** 依賴 Systick 中斷，`HAL_Init()` 必須先呼叫

當我們將上述 `main.c` 整合進 CMake 專案並嘗試編譯時，踩了一連串的坑。每一個錯誤背後都揭示了一個關於 STM32 生態系或交叉編譯的關鍵知識。以下是逐項記錄：

## 障礙 1：CMake 抓到原生編譯器而非 arm-none-eabi-gcc

**症狀**：`cmake ..` 成功，但 `make` 時報 `arm-none-eabi-gcc: 没有那个文件或目录`。

**原因**：CMake 在第一次執行時會快取編譯器路徑。如果執行 cmake 時 PATH 中沒有工具鏈目錄，CMake 會 fallback 到系統原生 gcc（`/usr/bin/cc`）。即使後續 PATH 已包含工具鏈，CMake 也不會重新檢測。

**解決**：執行 cmake 時必須保證 PATH 包含工具鏈，並用 `-DCMAKE_C_COMPILER` 明確指定編譯器名稱：

```bash
cd build
PATH="/path/to/arm-gnu-toolchain/bin:$PATH" cmake .. \
  -DCMAKE_C_COMPILER=arm-none-eabi-gcc \
  -DCMAKE_ASM_COMPILER=arm-none-eabi-gcc
```

編譯時同樣需要工具鏈在 PATH 中：
```bash
PATH="/path/to/arm-gnu-toolchain/bin:$PATH" make -j4
```

或者將工具鏈永久加入 `~/.bashrc`：
```bash
echo 'export PATH="/home/rock/stm32/arm-gnu-toolchain-15.2.rel1-aarch64-arm-none-eabi/bin:$PATH"' >> ~/.bashrc
```

**教訓**：交叉編譯環境中，cmake 和 make 兩階段都需要正確的 PATH。

## 障礙 2：裸機編譯器無法通過 CMake 的編譯器檢測

**症狀**：
```
The C compiler is not able to compile a simple test program.
...
undefined reference to `_exit'
undefined reference to `_close'
```
（注意：這是在執行 `cmake ..` 時發生的，不是 `make`。）

**原因**：CMake 在檢測編譯器時，會嘗試編譯並**連結**一個最小測試程式。但 `arm-none-eabi-gcc` 是裸機編譯器，沒有作業系統提供 `_exit`、`_close`、`_sbrk` 等 syscall 實作。在沒有連結腳本（`.ld`）和啟動檔（`.s`）輔助的情況下，它無法獨立完成連結。

**解決**：在 `CMakeLists.txt` 中、`project()` 指令**之前**加入：
```cmake
set(CMAKE_TRY_COMPILE_TARGET_TYPE STATIC_LIBRARY)
```
此後執行 cmake 時不需要再加 `-D` 參數。

**比較**：

| 方式 | 優點 | 缺點 |
|------|------|------|
| 寫在 `CMakeLists.txt` (推薦) | 一次設定，永久生效 | 每次新建專案都要手動加 |
| 指令列 `-DCMAKE_TRY_COMPILE_TARGET_TYPE=...` | 不修改 CMakeLists.txt | 每次 cmake 都要手動加，容易忘 |

**教訓**：嵌入式交叉編譯器與原生編譯器的差異——CMake 的預設檢測假設編譯器能產出可執行檔，但裸機工具鏈需要完整的連結描述才能辦到。把 `CMAKE_TRY_COMPILE_TARGET_TYPE` 寫入 CMakeLists.txt 是最穩妥的做法。

## 障礙 3：缺少 `STM32F103xB` 晶片型號巨集定義

**症狀**：
```
In file included from Core/system_stm32f1xx.c:58:
Drivers/CMSIS/Device/ST/STM32F1xx/Include/stm32f1xx.h:140: 
  #error "Please select first the target STM32F1xx device used in your application"
```
隨後也引發 HAL 原始檔中 `__IO`、`uint32_t` 等型別未定義的連鎖錯誤。

**原因**：STM32 的 CMSIS Device Header（`stm32f1xx.h`）是多晶片共用的。它透過預處理器巨集來決定要引入哪個晶片專屬的定義檔：
```c
#if defined(STM32F103xB)
  #include "stm32f103xb.h"   // ← 定義 __IO、暫存器位址、中斷編號等
#elif defined(STM32F103xE)
  #include "stm32f103xe.h"
...
#else
  #error "..."
#endif
```
如果沒有定義 `STM32F103xB`，所有類型與暫存器定義都是空的，編譯必然失敗。

**補充：error 從哪裡來？**
`main.c` 自己沒有直接 `#include "stm32f1xx.h"`，但專案中有其他原始檔會引入它：

```
system_stm32f1xx.c
  └─ #include "stm32f1xx.h"  ← 直接引用，第一個觸發 #error

stm32f1xx_hal_rcc.c
  └─ #include "stm32f1xx_hal_rcc.h"
       └─ #include "stm32f1xx_hal_def.h"
            └─ #include "stm32f1xx.h"  ← 間接引用

main.c
  └─ #include "stm32f1xx_hal.h"
       └─ (修正後) #include "stm32f1xx_hal_conf.h"
            └─ #include "stm32f1xx.h"  ← main.c 透過我們修好的 hal_conf.h 才引入
```

**解決**：在 CMake 編譯旗標中加入：
```cmake
set(CMAKE_C_FLAGS "${MCU_FLAGS} -DSTM32F103xB ...")
```
**教訓**：STM32 的 Device Header 依賴於一個「看不見的」巨集定義——這是 STM32Cube 生態系的慣例，但新手無從得知。這個錯誤通常最先在 `system_stm32f1xx.c` 觸發，因為它是專案中第一個直接 `#include "stm32f1xx.h"` 的檔案。

## 障礙 4：`stm32f1xx_hal.h` 不是「自足」的標頭檔

**症狀**：即使加了 `-DSTM32F103xB`，HAL 函數的類型（`HAL_StatusTypeDef`, `HAL_OK`）仍然未定義：
```
stm32f1xx_hal.h:282: error: unknown type name 'HAL_StatusTypeDef'
```

**原因**：這是本次建置中**最深、最難排查**的坑。追查 include 鏈可以發現：

| 檔案 | include 了什麼？ |
|------|-----------------|
| `main.c` | `#include "stm32f1xx_hal.h"` |
| `stm32f1xx_hal.h` (L29) | `#include "stm32f1xx_hal_conf.h"` **(唯一一行 include！)** |
| `stm32f1xx_hal_conf.h` | 你的客製配置（我們最初只有 `#define`，沒有 `#include`） |

問題來了：`stm32f1xx_hal.h` 在 L59 就使用了 `__IO uint32_t` 和 `HAL_StatusTypeDef`，但它**自己並沒有 include** 定義這些類型的檔案！

真正提供這些定義的是：
- `stm32f1xx_hal_def.h` → 定義 `HAL_StatusTypeDef`、`HAL_OK`、`assert_param`
- `stm32f1xx.h` → 透過 `stm32f103xb.h` + `core_cm3.h` → 定義 `__IO`、`uint32_t`

在 ST 官方的 STM32CubeMX 生成專案中，`stm32f1xx_hal_conf.h` 底部會 include 各個模組 header（如 `stm32f1xx_hal_rcc.h`、`stm32f1xx_hal_gpio.h`），而**這些模組 header 會各自 include `stm32f1xx_hal_def.h`**，從而形成完整的依賴鏈。

我們最初寫的 `hal_conf.h` 只有模組啟用巨集（`#define HAL_xxx_ENABLED`），沒有 include 任何 header，導致類型全部未定義。

**解決**：仿照 STM32Cube 範例專案的 `hal_conf.h` 寫法：
```c
#include "stm32f1xx.h"           // CMSIS device header
#include "stm32f1xx_hal_def.h"   // HAL 基礎類型定義

// ... 模組啟用巨集 ...

// module headers (必須放在啟用巨集之後)
#include "stm32f1xx_hal_rcc.h"
#include "stm32f1xx_hal_gpio.h"
#include "stm32f1xx_hal_cortex.h"
#include "stm32f1xx_hal_flash.h"
```

**教訓**：STM32 HAL 庫的 `stm32f1xx_hal.h` 並非自足（self-contained）標頭檔。它依賴 `stm32f1xx_hal_conf.h` 透過 include 其他模組 header 來間接補齊類型定義。這是 ST 的慣例設計，而非遺漏。

## 障礙 5：`#ifdef` vs `#if` 的 C 前置處理器陷阱

**症狀**：
```
undefined reference to `assert_failed'
dangerous relocation: unsupported relocation (ARM/Thumb)
```
到處都是 `assert_failed` 的未定義參考。

**原因**：我們在 `hal_conf.h` 中寫了：
```c
#define USE_FULL_ASSERT 0U

#ifdef USE_FULL_ASSERT
  #define assert_param(expr) ((expr) ? (void)0U : assert_failed(...))
#else
  #define assert_param(expr) ((void)0U)
#endif
```
但 `#ifdef` 只檢查巨集**是否被定義**，完全**不管其值為何**。`#define USE_FULL_ASSERT 0U` 使這個巨集被定義了，所以 `#ifdef` 判斷為真，即使值為 0，仍展開了 `assert_failed` 版本。

**補充：`assert_param` 到底是什麼？**
`assert_param()` 是 HAL 內部用來檢查函數參數合法性的機制。以 `HAL_GPIO_Init()` 為例，原始碼中會有：

```c
void HAL_GPIO_Init(GPIO_TypeDef *GPIOx, GPIO_InitTypeDef *GPIO_Init)
{
    assert_param(IS_GPIO_PIN(GPIO_Init->Pin));   // Pin 參數是否合法？
    assert_param(IS_GPIO_MODE(GPIO_Init->Mode));  // Mode 參數是否在允許範圍？
    // ... 實際初始化程式 ...
}
```

兩種模式的差異：

| 條件 | `assert_param` 展開為 | 效果 |
|------|----------------------|------|
| `#if 0` (關閉，預設) | `((void)0U)` | 什麼都不做，零開銷 |
| `#if 1` (開啟) | `(expr) ? (void)0U : assert_failed(__FILE__, __LINE__)` | 參數不合法時呼叫 `assert_failed()` |

啟用後，你必須自行實作 `assert_failed()`，例如：
```c
void assert_failed(uint8_t *file, uint32_t line)
{
    printf("Assert failed: file %s, line %ld\n", file, line);
    while(1);  // 停機
}
```
對正式 release 來說一般**關閉**（減少程式碼大小），僅在開發除錯時打開。

**解決**：改用 `#if`（檢查值）而非 `#ifdef`（檢查定義）：
```c
#if 0  /* 設為 1 啟用完整 assert 檢查 */
  #define assert_param(expr) ((expr) ? (void)0U : assert_failed(...))
#else
  #define assert_param(expr) ((void)0U)
#endif
```
**教訓**：`#ifdef` 與 `#if` 的差別是 C 語言常見陷阱。當你想用 0/1 開關功能時，永遠用 `#if`。

## 障礙 6：`Error_Handler()` 的隱式宣告

**症狀**：
```
main.c:53: error: implicit declaration of function 'Error_Handler'
main.c:90: warning: conflicting types for 'Error_Handler'
```

**原因**：`SystemClock_Config()` 在 L53 呼叫了 `Error_Handler()`，但該函數的定義在 L90（檔案尾部）。在 C99 標準中，函數在第一次使用前如果沒有前向宣告（forward declaration），編譯器會假設它回傳 `int`，這與實際的 `void` 回傳型別衝突。

**解決**：在檔案頂部 `static void GPIO_Init(void);` 之後加入：
```c
void Error_Handler(void);
```
**教訓**：這是 C 語言的基本規則，但在嵌入式專案中特別容易被忽略——因為在 IDE 中檔案結構通常是 IDE 自動生成的，開發者不太注意順序。

## 障礙 7：工具鏈不在 PATH 中——反覆出現

**症狀**：
```
make[2]: arm-none-eabi-gcc: 没有那个文件或目录
```
即使 cmake 配置成功，make 執行時仍然找不到編譯器。

**原因**：cmake 在配置階段透過 `-DCMAKE_C_COMPILER=arm-none-eabi-gcc` 記錄了編譯器名稱。但 make 在實際調用編譯器時，仍依賴 SHELL 的 PATH 環境變數來解析 `arm-none-eabi-gcc` 這個命令。如果 make 執行時的 PATH 沒有工具鏈目錄，就會失敗。

**解決**：每次執行 make 時都要確保 PATH 正確：
```bash
PATH="/home/rock/stm32/arm-gnu-toolchain-15.2.rel1-aarch64-arm-none-eabi/bin:$PATH" make -j4
```
或將工具鏈永久加入 `~/.bashrc`：
```bash
echo 'export PATH="/home/rock/stm32/arm-gnu-toolchain-15.2.rel1-aarch64-arm-none-eabi/bin:$PATH"' >> ~/.bashrc
```

## 🔥 燒錄與除錯：OpenOCD + ST-Link

編譯出 `main.bin` 後，透過 ST-Link 燒錄到 Blue Pill：

### 方式一：`make flash`（自動化）

`CMakeLists.txt` 已內建燒錄目標，直接在 `build/` 目錄執行：

```bash
# 建置 + 燒錄一次完成
PATH="/path/to/arm-gnu-toolchain/bin:$PATH" make flash

# 或用 .bin 檔燒錄
PATH="/path/to/arm-gnu-toolchain/bin:$PATH" make flash-bin
```

### 方式二：手動 OpenOCD 指令

```bash
# 燒錄 ELF（含除錯符號）
openocd -f interface/stlink.cfg -f target/stm32f1x.cfg \
  -c "program build/main.elf verify reset exit"

# 燒錄 raw binary（指定 flash 起始位址 0x08000000）
openocd -f interface/stlink.cfg -f target/stm32f1x.cfg \
  -c "program build/main.bin 0x08000000 verify reset exit"
```

#### 參數說明

| 參數 | 意義 |
|------|------|
| `-f interface/stlink.cfg` | 指定燒錄器為 ST-Link（V2/V3 通用） |
| `-f target/stm32f1x.cfg` | 指定目標晶片為 STM32F1 系列 |
| `program ... verify reset exit` | 燒錄 → 驗證 → 復位晶片 → 退出 |
| `0x08000000` | STM32F103 的 Flash 起始位址 |

### 互動除錯（GDB）

先啟動 OpenOCD 伺服器（背景執行）：

```bash
openocd -f interface/stlink.cfg -f target/stm32f1x.cfg &
```

再用 GDB 連線除錯：

```bash
PATH="/path/to/arm-gnu-toolchain/bin:$PATH" arm-none-eabi-gdb \
  -ex "target extended-remote :3333" \
  -ex "load" \
  -ex "monitor reset halt" \
  build/main.elf
```

或在 CMake 中直接啟動 OpenOCD 伺服器（不需手動輸入指令）：

```bash
make openocd
```

然後另開一個終端機連 GDB：
```bash
arm-none-eabi-gdb -ex "target extended-remote :3333" build/main.elf
```

## 🧹 補充：關於 `_read`、`_write` 的連結警告

在 `make` 連結階段你可能會看到這類訊息：
```
warning: _read is not implemented and will always fail
warning: _write is not implemented and will always fail
warning: _close is not implemented and will always fail
```

這些是**警告，不是錯誤**。它們來自 newlib 的 stub（空殼實作），對點燈程式完全沒有影響。原因是：

- `arm-none-eabi-gcc` 預設使用 newlib 的 C 標準函式庫
- newlib 需要底層的 `_read()`、`_write()`、`_close()` 等 syscall
- 裸機平台上沒有作業系統提供這些函數
- `-specs=nosys.specs`（已在 `CMakeLists.txt` 中）告訴連結器使用「總是失敗」的 stub 版本

要消除這些警告，可以自己實作這幾個 syscall（例如用 UART 實作 `_write` 來輸出到串列埠），但對點燈範例來說完全不需要。

## 總結：一個點燈專案的技術帳單

| # | 障礙 | 領域 | 根本原因 |
|---|------|------|---------|
| 1 | CMake 抓到錯誤編譯器 | 建置系統 | PATH 管理不當 |
| 2 | CMake 檢測失敗 | 建置系統 | 裸機工具鏈無法獨立連結 |
| 3 | 缺少晶片型號巨集 | CMSIS/Cube 慣例 | `-DSTM32F103xB` 是「隱形」的必要 define |
| 4 | HAL header 非自足 | STM32Cube 架構 | `stm32f1xx_hal_conf.h` 缺少關鍵 include 鏈 |
| 5 | assert_failed 未定義 | C 預處理器 | `#ifdef` vs `#if` 誤用 |
| 6 | Error_Handler 隱式宣告 | C 語言規範 | 缺少前向宣告 |
| 7 | 工具鏈不在 PATH | 環境配置 | cmake/make 兩階段都要 PATH |

這 7 個障礙中，**障礙 4（HAL header include 鏈）**耗時最長、最難排查。它暴露了 STM32Cube 生態系的一個核心設計特點：`stm32f1xx_hal.h` 並非自足標頭檔，它依賴 `stm32f1xx_hal_conf.h` 透過 include 模組 header 來間接完成完整類型定義。這個慣例對 STM32CubeMX 用戶是完全透明的（因為工具自動生成），但對手動建構專案的人來說，卻是一個需要深入理解的知識點。
